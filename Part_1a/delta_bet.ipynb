```python
from math import * 
import re, sys
import numpy as np
import matplotlib.pyplot as plt
# esp = 1e-3
```

```python
def exp_utility(x, bliss =1, m=1, r = 7e-2):
    return bliss - m * exp(- r * x)

def inv_exp_utility(u, bliss =1, m=1, r = 7e-2):
    return -log(bliss - u / m)/r

exp_utility(0)
```

```python
money = np.arange(100)
plt.plot(money, [exp_utility(z) for z in money])
```

```python
 u = np.linspace(0,0.99)
plt.plot(u, [inv_exp_utility(z) for z in u], 'r')
```

```python
[inv_exp_utility(exp_utility(x)) for x in money]
```

```python
# vary the limits of a even bet and compute the certain equivalent

delta_bet = np.logspace(log10(1), log10(50))

d =0
def certain_equiv(d):
     return inv_exp_utility(0.5 * exp_utility(-d) + 0.5 * exp_utility(d))
[certain_equiv(d) for d in delta_bet]
```
