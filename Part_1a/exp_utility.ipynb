Use the participants investment fraction estimates to estimate their risk aversion parameter, $r$. Assume their $r$ is independent of wealth, using
an exponential utility model $u(v) = -e^{-v}$.

```python
from math import cosh, exp, pow
import matplotlib.pyplot as plt

def constant_r_utility(v, r):
    '1e-3 - almost risk neutral, 0.5 - maxes at 100,000 '
    s = r * 1e-4
    return 1e6 *(1- exp(-s * v))
```

```python

v = range(0, int(1E6), 5000)
u = [constant_r_utility(z, 5e-2) for z in v]
plt.plot(v,u)
```

## An exponential utility function

$ u(v,w) = w(2 - \exp(av/w)$)

where increasing $a$ incurs increasing risk premium.  (What are it's units?)

The utility risk premium:

$risk(v,w ) = w (1 - \cosh (av/w))$

where $w$ is wealth and $v$ is defines the lottery $p=1/2$, $p(+v) + p(-v)$ 

```python
MAX_W = 1e6   # A precaution to prevent exp() overflow
def exp_risk_premium(v, w, a):
    'Problem to keep cosh in range'
    if w > MAX_W:
        print('Increase w to at least {MAX_W}')
        return None
    return w * (1 - cosh(a*v* 1e-6))
```
